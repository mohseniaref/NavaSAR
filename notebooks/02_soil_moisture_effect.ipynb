# 02 · Soil Moisture Effect on Coherence and Phase

Reproduces **Fig. 3** and **Fig. 4** of De Zan et al. (2014):

- **Fig. 3** — Interferometric coherence magnitude as a 2-D function of master
  and slave soil moisture.
- **Fig. 4** — Interferometric phase (degrees) as a 2-D function of master and
  slave soil moisture.

Settings: L-band (1.4 GHz), 45° incidence, 51 % sand, 13 % clay.

---
**References**
- De Zan, F., Parizzi, A., Prats-Iraola, P., & López-Dekker, P. (2014).
  *A SAR interferometric model for soil moisture.*
  IEEE Transactions on Geoscience and Remote Sensing, 52(1), 418–425.
  https://doi.org/10.1109/TGRS.2013.2241069

```python
import sys
sys.path.insert(0, '..')

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator
from navasar.coherence import interferometric_coherence

plt.rcParams.update({'font.size': 11, 'figure.dpi': 120})

KW = dict(freq_ghz=1.4, theta_inc_deg=45.0, sand=0.51, clay=0.13)
```

## Build master × slave moisture grid

```python
mv = np.linspace(0.0, 0.5, 100)
MV1, MV2 = np.meshgrid(mv, mv)   # master, slave

gamma = interferometric_coherence(MV1, MV2, **KW)
coh_mag   = np.abs(gamma)
coh_phase = np.rad2deg(np.angle(gamma))
```

## Fig. 3 — Coherence magnitude (De Zan 2014)

```python
fig, ax = plt.subplots(figsize=(5.5, 4.5))
im = ax.contourf(mv, mv, coh_mag, levels=20, cmap='viridis', vmin=0, vmax=1)
cs = ax.contour(mv, mv, coh_mag, levels=10, colors='white', linewidths=0.5, alpha=0.5)
plt.colorbar(im, ax=ax, label='|γ|')
ax.set_xlabel('Master moisture $m_v$')
ax.set_ylabel('Slave moisture $m_v$')
ax.set_title('Interferometric coherence magnitude\n(De Zan 2014, Fig. 3)')
ax.xaxis.set_major_locator(MultipleLocator(0.1))
ax.yaxis.set_major_locator(MultipleLocator(0.1))
plt.tight_layout()
plt.savefig('../examples/fig03_coherence_magnitude.png', dpi=150)
plt.show()
```

## Fig. 4 — Interferometric phase (De Zan 2014)

```python
fig, ax = plt.subplots(figsize=(5.5, 4.5))
im = ax.contourf(mv, mv, coh_phase, levels=20, cmap='RdBu_r')
cs = ax.contour(mv, mv, coh_phase, levels=10, colors='k', linewidths=0.5, alpha=0.4)
ax.clabel(cs, fmt='%d°', fontsize=7)
plt.colorbar(im, ax=ax, label='Phase [deg]')
ax.set_xlabel('Master moisture $m_v$')
ax.set_ylabel('Slave moisture $m_v$')
ax.set_title('Interferometric phase [degrees]\n(De Zan 2014, Fig. 4)')
ax.xaxis.set_major_locator(MultipleLocator(0.1))
ax.yaxis.set_major_locator(MultipleLocator(0.1))
plt.tight_layout()
plt.savefig('../examples/fig04_interferometric_phase.png', dpi=150)
plt.show()
```

## Phase sensitivity: 20% → 21% moisture change

```python
g = interferometric_coherence(0.20, 0.21, **KW)
print(f'Phase change 20%→21% mv: {np.rad2deg(np.angle(g)):.1f} deg')
print(f'Coherence magnitude:      {abs(g):.4f}')
```

## Phase along diagonal slices (fixed master)

```python
fig, ax = plt.subplots(figsize=(6, 4))
for mv_master in [0.05, 0.15, 0.25, 0.35]:
    g_slice = interferometric_coherence(mv_master, mv, **KW)
    ax.plot(mv, np.rad2deg(np.angle(g_slice)), label=f'master={mv_master:.2f}')
ax.axhline(0, color='k', lw=0.8, ls='--')
ax.set_xlabel('Slave moisture $m_v$')
ax.set_ylabel('Interferometric phase [deg]')
ax.set_title('Phase vs slave moisture for fixed masters')
ax.legend(fontsize=9)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('../examples/fig04b_phase_slices.png', dpi=150)
plt.show()
```
